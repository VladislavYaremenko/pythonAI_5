```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
population = np.random.normal(
    loc=70,
    scale=10,
    size=50000
)
sample_sizes = [5, 30]
num_samples = 1000
means = {}

for n in sample_sizes:
    sample_means = []

    for i in range(num_samples):
        sample = np.random.choice(population, size=n, replace=False)
        sample_means.append(np.mean(sample))

    means[n] = np.array(sample_means)
plt.figure(figsize=(12, 5))

for i, n in enumerate(sample_sizes, 1):
    plt.subplot(1, 2, i)
    plt.hist(means[n], bins=30, edgecolor="black")
    plt.title(f"Вибіркові середні, n = {n}")
    plt.xlabel("Середнє значення")
    plt.ylabel("Частота")
    plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
for n in sample_sizes:
    print(f"\nРозмір підвибірки n = {n}")
    print("Середнє вибіркових середніх:", np.mean(means[n]))
    print("Стандартне відхилення вибіркових середніх:", np.std(means[n]))
```
